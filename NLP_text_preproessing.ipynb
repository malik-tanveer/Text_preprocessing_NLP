## Rules of the Text Preprocessing
- Lower text
- Removing Space
- removing numbers
- Handle Punctuation
- Steeming
- Stopwords

```python
# Function of the text Preprocessing

def text_preprocessing(text):
  text = text.lower()
  text = re.sub("\s+", " ", text)
  text = re.sub("\d+", " ", text)
  punctuations = "".join([a for a in text if a not in string.punctuation])
  text = " ".join([a for a in punctuations.split() if a not in stopwords.words("english")])

  return text
```

```python
# Map Function to handle a sentiment values
data["sentiment"] = data["sentiment"].map({
    "Positive": 1,
    "Neutral": 0,
    "Negative": -1,
    "Irrelevant": 2
})
```

```python
data["sentiment"].unique()
```

```python
# Apply on the Reviews dataset
data["reviews"] = data["reviews"].apply(text_preprocessing)
```

```python
tok = Tokenizer()
```

```python
data["reviews"] = tok.texts_to_sequences(data["reviews"])
```
